# Clustering: O que é Cluster Analysis e quais suas Aplicações?
(https://www.datageeks.com.br/clustering/)

O **clustering**, ou agrupamento, é uma técnica essencial para **machine learning** e **ciência de dados**, especialmente relevante em contextos onde a identificação de padrões intrínsecos dentro de conjuntos de dados não rotulados é crucial.

Diferentemente de métodos supervisionados, onde os modelos são treinados com dados rotulados, o clustering pertence à classe dos **métodos de aprendizado não supervisionado**, onde a estrutura dos dados não é conhecida a priori. Essa abordagem exploratória permite identificar grupos similares de objetos com base em suas características e atributos comuns.

### Explique os dois passos que o K-Means repete: distribuir pontos e atualizar centros.
• **Distribuir os pontos (Atribuição)**: O algoritmo calcula a distância de cada ponto dos dados em relação a todos os centroides disponíveis. Cada ponto é associado ao centroide que estiver mais perto dele, formando os grupos iniciais.

• **Atualizar os centros (Atualização)**: Com os grupos formados, o algoritmo calcula a média geométrica (a posição central) de todos os pontos que ficaram dentro de cada grupo. O centroide é então movido para essa nova posição central.

Esses passos se repetem (o centro muda, os pontos mudam de grupo, o centro muda de novo) até que os centroides parem de se mover.

### O que é um centroide? Ele precisa ser um dos pontos originais?
O centroide é o **centro de gravidade** ou a **média geométrica de um grupo de pontos**. Pense nele como o "ponto central ideal" que resume a localização daquele grupo.

• Ele precisa ser um dos pontos originais? Não. Como o centroide é calculado pela média das coordenadas de vários pontos, **ele quase sempre será uma coordenada inédita no espaço**, flutuando no meio do grupo, e não um dado real que já existia no seu dataset.

### Por que o K-Means é chamado de aprendizado não supervisionado? Recebemos grupos corretos como resposta durante o treinamento?
O K-Means é chamado de aprendizado não supervisionado porque ele **trabalha com dados que não possuem etiquetas (labels) ou respostas corretas pré-definidas**. Ele precisa encontrar padrões, semelhanças e estruturas nos dados sozinho, apenas olhando para a distância entre eles.

• Recebemos grupos corretos durante o treinamento? Não. Não existe um "gabarito". O algoritmo não sabe o que cada grupo significa (por exemplo, ele não sabe que o Grupo A são "clientes VIP" e o Grupo B são "clientes casuais"). Ele apenas agrupa os dados por proximidade, cabendo ao ser humano interpretar o significado de cada grupo depois.
