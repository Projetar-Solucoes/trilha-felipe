```python
import pandas as pd
def mensagem(mensagem):
  print(mensagem)
def vazio():
  print()

df = pd.read_csv("https://raw.githubusercontent.com/Projetar-Solucoes/materiais-trilha/main/semana-04/solicitacoes_semana4.csv")
mensagem('Visualizar Primeiras 5 Linhas')
display(df.head()) #visualiza primeiras 5 linhas
vazio()

mensagem("Mostrar Quantas Linhas/Colunas")
display(df.shape) #mostra quantas (linhas,colunas)
vazio()

mensagem("Como o Pandas Interpreta cada Dado em cada Coluna")
display(df.dtypes) #como o pandas interpreta cada dado
vazio()

mensagem("Calcula Quantas Colunas Vazias tem pelo .isna retornando True ou False, e o .sum somando os True")
display(df.isna().sum()) #isna retorna True ou False se está vazio, sum soma aqueles True vazio
vazio()

mensagem('Calcula Quantas Linhas Repetidas Tem pelo .duplicated retornando True ou False, e o .sum somando os True')
display(df.duplicated().sum()) #duplicated verifica repetidos, retorna True ou False, e o sum soma os True duplicado
vazio()

mensagem('Mostra as Estatísticas da Categoria e soma-as pelo .value_counts')
print(df["categoria"].value_counts())
vazio()

mensagem("Mostra as Estatísticas da Prioridade e soma-as pelo .value_counts")
print(df["prioridade"].value_counts())
vazio()

mensagem("Mostra as Estatísticas do Status e soma-as pelo .value_counts")
print(df["status"].value_counts())
vazio()

mensagem("Mostra as Estatísticas da Unidade e soma-as pelo .value_counts")
print(df["unidade"].value_counts())
vazio()
```
