### Importando o pandas

```python
import pandas as pd
```

### Criando um dataframe a partir de um dicionário

```python
# dataframe = pd.DataFrame()
venda = {'data': ['15/02/2021', '16/02/2021'],
         'valor': [500, 300],
         'produto': ['feijao', 'arroz'],
         'qtde': [50, 70],
        }
vendas_df = pd.DataFrame(venda)
```

### Visualização dos Dados
    - print
    - display

```python
display(vendas_df)
```

### Importando arquivos e bases de dados

```python
vendas_df = pd.read_excel("Vendas.xlsx") #ler o excel como tabela pandas, posso colocar o caminho também
display(vendas_df)
```

### Resumos de Visualização de Dados simples e úteis
    - head
    - shape
    - describe

```python
display(vendas_df.head(10)) #só mostra as linhas que eu quiser, nesse caso, 10 primeiras
print(vendas_df.shape) #mostra quantas linhas e colunas a tabela tem
display(vendas_df.describe()) #estatísticas
```

### Pegar 1 coluna (e os pd.Series = Coluna ou Linha do Pandas:)

```python
produtos = vendas_df[['Produto', 'ID Loja']] #1 coluna vem desistilizado, mas mais de 1 vem como tabela estilizada
display(produtos)
```

### .loc, um método muito importante
    - Pegar 1 linha
    - Pegar linhas de acordo com alguma condição
    - Pegar linhas e colunas específicas
    - Pegar 1 valor específico

```python
# pegar uma linha
display(vendas_df.loc[1:5])

# pegar linhas que correspondem a uma condição
vendas_norteshopping_df = vendas_df.loc[vendas_df['ID Loja'] == 'Norte Shopping']

# pegar várias linhas e colunas usando o loc
vendas_norteshopping_df = vendas_df.loc[vendas_df['ID Loja'] == 'Norte Shopping', ["ID Loja", "Produto", "Quantidade"]]
display(vendas_norteshopping_df)

# pegar 1 valor específico
print(vendas_df.loc[1, 'Produto'])
#loc[linhas, colunas]
```

### Adicionar 1 coluna

```python
# a partir de uma coluna que existe
vendas_df['Comissão'] = vendas_df['Valor Final'] * 0.05
# display(vendas_df)

# criar uma coluna com valor padrão
vendas_df.loc[:, "Imposto"] = 0
display(vendas_df)
#usa o loc pra adicionar sem dar muito problema, [todas as linhas = :, coluna]
```

### Adicionar 1 linha
    - Linhas de um complemento da base de dados

```python
vendas_dez_df = pd.read_excel("Vendas - Dez.xlsx")
#display(vendas_dez_df)

vendas_df = vendas_df.append(vendas_dez_df)
display(vendas_df)
#adiciona as linhas de dezembro na tabela
```

### Excluir linhas e colunas

```python
vendas_df = vendas_df.drop(0, axis=0) #axis(eixo): 1=coluna, 0=linha
display(vendas_df)
```

### Valores Vazios
    - Deletar linhas/colunas vazias
    - Deletar linhas que possuem valores vazios
    - Preencher valores vazios (média e último valor)

```python
# deletar linhas e colunas completamente vazias
#vendas_df = vendas_df.dropna(how='all', axis=1)

# deletar linhas que possuem pelo menos 1 valor vazio
#vendas_df = vendas_df.dropna()

# preencher valores vazios
# preencher com a média da coluna
vendas_df['Comissão'] = vendas_df['Comissão'].fillna(vendas_df['Comissão'].mean())
display(vendas_df)

# preencher com o último valor
vendas_df = vendas_df.ffill()
```

### Calcular Indicadores
    - Groupby
    - Value Counts

```python
# value counts
transacoes_loja = vendas_df['ID Loja'].value_counts()
display(transacoes_loja)

# group by
faturamento_produto = vendas_df[['Produto', 'Valor Final']].groupby('Produto').sum()
display(faturamento_produto)
```

### Mesclar 2 dataframes (Procurar informações de um dataframe em outro)

```python
gerentes_df = pd.read_excel("Gerentes.xlsx")
# display(gerentes_df)

vendas_df = vendas_df.merge(gerentes_df)
display(vendas_df)
```

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/Projetar-Solucoes/materiais-trilha/main/semana-04/solicitacoes_semana4.csv")
display(df)
```

### Verificando linhas duplicadas
Podemos usar o método `.duplicated()` que retorna um valor booleano (`True` para duplicadas e `False` para únicas) e somar os valores com `.sum()` para descobrir o total de linhas repetidas.

```python
# Conta a quantidade total de linhas duplicadas (considerando todas as colunas)
total_duplicadas = df.duplicated().sum()
print(f"Total de linhas completamente duplicadas: {total_duplicadas}")
```

Caso queira visualizar quais são essas linhas duplicadas na tela, podemos filtrar o DataFrame da seguinte forma:

```python
# Exibe as linhas duplicadas (mantém a primeira ocorrência e mostra apenas as repetições)
linhas_duplicadas_df = df[df.duplicated()]
display(linhas_duplicadas_df)
```

### Verificando valores vazios na coluna prioridade
Podemos filtrar a coluna desejada e usar `.isnull().sum()` para contar as linhas sem preenchimento.

```python
# Conta a quantidade de valores vazios na coluna 'prioridade'
vazios_prioridade = df['prioridade'].isnull().sum()
print(f"Quantidade de valores vazios na coluna prioridade: {vazios_prioridade}")
```

### Verificando valores únicos na coluna prioridade
Podemos usar `.nunique()` para contar os valores distintos (ignorando valores nulos por padrão) e `.unique()` para listar quais são eles.

```python
# Conta a quantidade de valores diferentes (únicos)
quantidade_unicos = df['categoria'].nunique()

# Obtém os valores diferentes
valores_unicos = df['categoria'].unique()

print(f"Quantidade de valores diferentes na coluna prioridade: {quantidade_unicos}")
print(f"Valores existentes na coluna: {valores_unicos}")
```
