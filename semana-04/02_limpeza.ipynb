```python
# Passo 1: Remover as linhas repetidas
df = df.drop_duplicates().copy()

# Exibe quantas linhas restaram
print(f"Após remover as linhas repetidas, restaram {df.shape[0]} linhas.")
```

```python
# Passo 2: Deixar a categoria num padrão só
df["categoria"] = df["categoria"].str.strip().str.lower()

padrao = {
    "suporte ti": "Suporte TI", "suporte de ti": "Suporte TI",
    "acesso": "Acesso", "financeiro": "Financeiro",
    "infraestrutura": "Infraestrutura", "infra": "Infraestrutura",
    "dúvida": "Dúvida", "duvida": "Dúvida",
}

df["categoria"] = df["categoria"].map(padrao)

# Confere se ficou algum valor nulo após o mapeamento
print("Valores nulos na categoria:", df["categoria"].isna().sum())

# Mostra a quantidade de cada categoria
print("\nQuantidade por categoria:")
print(df["categoria"].value_counts())
```

```python
# Passo 3: Ajustar a primeira letra do status para maiúscula
df["status"] = df["status"].str.capitalize()

# Exibe o resultado das categorias de status
print(df["status"].value_counts())
```

```python
# Passo 4: Filtrar e contar avaliações válidas (entre 1 e 5)
avaliacoes_validas = df["avaliacao"].dropna()
avaliacoes_no_intervalo = avaliacoes_validas[(avaliacoes_validas >= 1) & (avaliacoes_validas <= 5)]

print(f"Total de avaliações válidas (notas de 1 a 5): {len(avaliacoes_no_intervalo)}")
```

# **Tratamento da Prioridade e Validação Final**

```python
# Preenchendo valores ausentes da prioridade com 'Não Informada' ou mantendo o padrão
df['prioridade'] = df['prioridade'].fillna('Não Informada')
print("Estatísticas de Prioridade após tratamento:")
print(df['prioridade'].value_counts())
```

### **Validação do Total de Linhas**

A soma das quantidades de solicitações por categoria deve bater exatamente com as 240 linhas da base limpa.

```python
soma_categorias = df["categoria"].value_counts().sum()
total_linhas = df.shape[0]

print(f"Soma das categorias: {soma_categorias}")
print(f"Total de linhas da base limpa: {total_linhas}")
print(f"A validação bateu? {soma_categorias == total_linhas}")
```

### **Média de Avaliação por Categoria (Apenas Notas de 1 a 5)**

```python
# Criamos uma coluna temporária contendo apenas notas de 1 a 5 para fins de cálculo de média
df_temp = df.copy()
df_temp['avaliacao_filtrada'] = df_temp['avaliacao'].apply(lambda x: x if (x >= 1 and x <= 5) else None)

print("Média de avaliação por categoria (apenas notas válidas de 1 a 5):")
display(df_temp.groupby("categoria")["avaliacao_filtrada"].mean())
```
